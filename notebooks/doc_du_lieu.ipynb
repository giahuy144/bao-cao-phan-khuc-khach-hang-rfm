```python
from pathlib import Path
from pyspark.sql import SparkSession

ROOT = Path.cwd().parent if Path.cwd().name == "notebooks" else Path.cwd()
CLEAN = ROOT / "data" / "clean"

spark = (
    SparkSession.builder
    .appName("RFM-view-clean")
    .master("local[*]")
    .config("spark.driver.memory", "4g")
    .getOrCreate()
)

article_df      = spark.read.parquet(str(CLEAN / "article_clean.parquet"))
customer_df     = spark.read.parquet(str(CLEAN / "customer_clean.parquet"))
transactions_df = spark.read.parquet(str(CLEAN / "transactions_clean.parquet"))
```

```python
# bảng article
print("Số dòng:", article_df.count())
print("Số cột :", len(article_df.columns))
article_df.printSchema()
article_df.show(10, truncate=False)

article_df.limit(10).toPandas()   # hiển thị dạng bảng đẹp trong Jupyter
```

```python
# bảng customer
print("Số dòng:", customer_df.count())
print("Số cột :", len(customer_df.columns))
customer_df.printSchema()
customer_df.show(10, truncate=False)

customer_df.limit(10).toPandas()
```

```python
# bảng transactions
print("Số dòng:", transactions_df.count())
print("Số cột :", len(transactions_df.columns))
transactions_df.printSchema()
transactions_df.show(10, truncate=False)

transactions_df.limit(10).toPandas()
```
